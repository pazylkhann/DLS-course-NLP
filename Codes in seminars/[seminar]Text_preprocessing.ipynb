```python
# StopWords and preprocessor(Tfidf)
from nltk.corpus import stopwords
StopWords = set(stopwords.words('english'))
nltk.download('wordnet')
wnl = nltk.WordNetLemmatizer()

def preproc_nltk(text):
  return ' '.join([wnl.lemmatize(word) for word in word_tokenize(text.lower()) if word not in StopWords])   # Preprocessor function of text using WordNetLemmatizer
st = "Oh, I think I ve landed Where there are miracles at work,  For the thirst and for the hunger Come the conference of birds"
preproc_nltk(st)
```

```python
%%time
vectorizer = TfidfVectorizer(preprocessor=preproc_nltk)  # vectorizer Tfidf which uses our preproc_nltk function for preprocessing
vectors = vectorizer.fit_transform(newsgroups_train.data)  # vectorizing text by the fit_transform() attribute(function)
```

```python
# preprocessor_spaCy
nlp = spacy.load('en_core_web_sm')
texts = newsgroups_train.data.copy()

def preproc_spacy(text):
  spacy_results = nlp(text)
  return ' '.join(token.lemma_ for token in spacy_results if token.lemma_ not in StopWords)
preproc_spacy(st)
```

```python
%%time
new_texts = []
for doc in nlp.pipe(texts, batch_size=32, n_process=3, disable=["parser", "ner"]):           # It's recommended by spaCy when working with a huge amount of text
  new_texts.append(' '.join(token.lemma_ for token in doc if token.lemma_ not in StopWords)) # We should use it by Pipelining it in such way
vectorizer = TfidfVectorizer()                                                               # So text can be vectorized Faster
vectors = vectorizer.fit_transform(new_texts)
```

###Let's look at the cosine measure between vectors

```python
import numpy as np
from numpy.linalg import norm

type(vectors)
```

```python
vector = vectors.todense()[0]
(vector != 0).sum()
```

```python
dense_vectors = vectors.todense()
dense_vectors.shape
```

```python
def cosine_sim(v1, v2):
  # v1, v2 (1 x dim)
  return np.array(v1 @ v2.T / norm(v1) / norm(v2))[0][0]
```

```python
cosine_sim(dense_vectors[0], dense_vectors[0])
```

```python
cosines = []
for i in range(10):
  cosines.append(cosine_sim(dense_vectors[0], dense_vectors[i]))
```

##Embeddings

```python
!pip install gensim
import gensim.downloader as api
embeddings_pretrained = api.load('glove-twitter-25')
```

```python
from gensim.models import Word2Vec

proc_words = [preproc_nltk(text).split() for text in newsgroups_train.data]
embeddings_trained = Word2Vec(proc_words,  # data for model to train on
                              vector_size=32,     # embedding vector size
                              min_count=5, # consider words that occured at least 5 times
                              window=3)
```

```python
def vectorize_sum(comment, embeddings):
  """
  Implement a function that converts preprocessed comment to token vectors
  """
  embeddings_dim = embeddings.vector_size
  features = np.zeros([embeddings_dim], dtype='float32')

  for word in preproc_nltk(comment).split():
    if word in embeddings:
      features += embeddings[word]
  return features
```

```python
X_wv = np.stack([vectorize_sum(text, embeddings_pretrained) for text in newsgroups_train.data])
X_train_wv, X_test_wv, y_train, y_test = train_test_split(X_wv, newsgroups_train.target, test_size=0.2, random_state=0)
X_train_wv.shape, X_test_wv.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

clf = LogisticRegression(max_iter=10000)
wv_model = clf.fit(X_train_wv, y_train)
roc_auc_score(y_test, wv_model.predict_proba(X_test_wv), multi_class='ovr')
```

```python
accuracy_score(y_test, wv_model.predict(X_test_wv))
```
